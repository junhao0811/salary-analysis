```python
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
# Filter the salaries DataFrame to include only Full-time (FT) employment type
full_time = salaries[salaries['employment_type'] == 'FT']

# Compute the highest salary in USD
highest_salary = {'salary_in_usd': 'max'}

# Perform the aggregation and group them by 'job_title'
groupbyJobTitle = full_time.groupby('job_title').agg(highest_salary).reset_index()

# Sets the size of plot
ax = groupbyJobTitle.plot.bar(figsize = (20, 5))

# Use values of column 'job_title' as the x-axis labels
ax.set_xticklabels(groupbyJobTitle['job_title'], rotation = 90)

# Setting a label for x-axis
plt.xlabel('Job Title')

# Setting a label for y-axis
plt.ylabel('Salary in USD')

# Setting the title of chart
plt.title('Job Salaries for Full-time (FT) employment type')
```

Answer: The bar chart above displays the job salaries for Full-time (FT) employment type. Based on the visualisation above, we can see that Research Scientist has the highest salary for Full-time (FT) employment type. One main reason is because of the specialization and experience that a Research Scientist requires. Full-time research scientists usually have many years of experience and specialized skills that are highly sought after. There are also very high demand for full-time research scientists because their work is highly sought after by many companies. Moreover, the supply of full-time research scientists are extremely limited. At last, their work can be very tedious when it comes to committing full-time research, so they can be very stressful when it comes to work sometimes. Hence, this explains why research scientist has the highest salary for Full-time (FT) employment type.

```python
# Put the experience level in the correct order
salaries['experience_level'] = pd.Categorical(salaries['experience_level'], categories=['EN','MI','SE','EX'])

# Find out the association between Experience Level and the Salary for all jobs
salaries.boxplot(column='salary_in_usd', by='experience_level')

# Setting limit on y-axis
plt.ylim(0, 460000)

# Setting a label for x-axis and y-axis
plt.xlabel('Experience Level')
plt.ylabel('Salary')
```

Answer: From the box plot above, we can clearly see that there is an association between Experience Level and the Salary for all jobs. We can see that entry-level has the lowest median salary among all other experience levels, and for executive-level it has the highest median salary among all. As the experience level of the employee increases, the median salary also increase. Thus, it exists a positive correlation between them.

B3.2:

Code:

```python
# Convert 'experience_level' to numbers
exp_lvl = {'EN': 1, 'MI': 2, 'SE': 3, 'EX': 4}
salaries['exp_lvl_by_num'] = salaries['experience_level'].map(exp_lvl)

# Create a new dataframe with the columns required
filt_columns = ['job_title', 'exp_lvl_by_num', 'salary_in_usd']
newDf = salaries[filt_columns]

# Group the dataframe by 'job_title'
all_jobs_df = newDf.groupby('job_title')

# Create two empty lists to store the data
job_title = []
correlation_coefficient = []

# For each group, calculate the correlation between 'exp_lvl_by_num' and 'salary_in_usd' for each job title
for job, data in all_jobs_df:
    
    # Only calculate the correlation if the job has all 4 types of experience level
    if data['exp_lvl_by_num'].nunique() == 4:
        correlation_coefficient_value = data['exp_lvl_by_num'].corr(data['salary_in_usd'])

        # Only include the correlation coefficient value with positive correlation
        if not correlation_coefficient_value <= 0:
            # Then add the job title and correlation above to their respective list
            job_title.append(job)
            correlation_coefficient.append(round(correlation_coefficient_value, 1))

# Create a new dataframe to plot the bar chart later on
correlation_df = pd.DataFrame({'Job Title': job_title, 'Correlation Coefficient': correlation_coefficient})

# Sets the size of plot
ax = correlation_df.plot.bar(figsize = (20, 7))

# Use values of column 'job_title' as the x-axis labels
ax.set_xticklabels(correlation_df['Job Title'], rotation = 45)

# Setting a label for x-axis
plt.xlabel('Job Title')

# Setting a label for y-axis
plt.ylabel('Correlation Coefficient')

# Setting the title of chart
plt.title('Association between Experience Level and Salary by Job Title')
```

Answer: The job with the highest association between Experience Level and the Salary is BI Data Analyst. From the bar chart above, we can see that this job has a positive correlation of 0.7 between different experience level and salary in USD, which appears to be the highest in the bar chart above. It shows that the salary of this job is highly based on how much experience does the employee has. For example, if the experience level of an employee jumps from senior-level to executive-level, their salary will have a high chance to receive a significant increase by the company they are working for.
